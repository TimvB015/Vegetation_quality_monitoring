#### Symbol definitions

- $N$: total number of samples.
- $C$: number of classes.
- $y_i$: true (ground-truth) label of sample $i$.
- $\hat{y}_i$: predicted label of sample $i$.
- $\mathbf{1}(\hat{y}_i=y_i)$: indicator function; equals 1 if the prediction is correct, else 0.

Per class $c$ (one-vs-rest view):
- $TP_c$ (true positives): number of samples with true class $c$ predicted as $c$.
- $FP_c$ (false positives): number of samples with true class $\neq c$ predicted as $c$.
- $FN_c$ (false negatives): number of samples with true class $c$ predicted as $\neq c$.
- $P_c$: precision for class $c$.
- $R_c$: recall for class $c$.
- $F1_c$: F1-score for class $c$.

Weights for weighted F1:
- $n_c$: support of class $c$ (number of samples whose true label is $c$).
- $\frac{n_c}{N}$: class weight used in the weighted average.

### F1 explanation

_The F1-score is the harmonic mean of Precision and recall. It's high only when both precision (few false positives) and Recall (few false negatives) are high._

$$
F1 = \frac{2PR}{P+R}
$$

where

$$
P=\frac{TP}{TP+FP}, \qquad R=\frac{TP}{TP+FN}
$$

Equivalent form:

$$
F1=\frac{2TP}{2TP+FP+FN}
$$

### Accuracy
$$
\text{Accuracy}=\frac{\sum_{i=1}^{N}\mathbf{1}(\hat{y}_i=y_i)}{N}
$$

Equivalently (multiclass):
$$
\text{Accuracy}=\frac{\sum_{c=1}^{C} TP_c}{N}
$$

### Precision / Recall (per class c)
$$
P_c=\frac{TP_c}{TP_c+FP_c}
\qquad
R_c=\frac{TP_c}{TP_c+FN_c}
$$

### F1 (per class c)
$$
F1_c=\frac{2P_cR_c}{P_c+R_c}
=\frac{2TP_c}{2TP_c+FP_c+FN_c}
$$

### Macro F1
$$
F1_{\text{macro}}=\frac{1}{C}\sum_{c=1}^{C} F1_c
$$

### Weighted F1
Let $n_c$ be the support (number of true samples) for class $c$, and $N=\sum_{c=1}^{C} n_c$.
$$
F1_{\text{weighted}}=\sum_{c=1}^{C}\left(\frac{n_c}{N}\right)F1_c
$$
