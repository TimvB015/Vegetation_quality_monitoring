## Thought process described

Often (normalized)-Entropy would be a good decisive parameter. But in this instance a perfect 3 fold division would be rather be described by random/unstable than a tie. Therefore a lader was developed: <br>
-> 80-100% True classification <br>
<!-- -> 70-80% Med classification <br> -->
-> >=60-80% Weak classification <br>
-> 2x 40% classification is a tie <br>
-> remaining is based on entropy <br>

This creates that a 4/10 + 4/10 + 2/10 is on the border of a tie (shows signs of a starting transition). Where 6/9 is just weakly classified, one can see a transition happening, but it is really not a tie yet.
