```python
from functions.gpkg_funcs import (
    import_gpkg_func,
)

from notebooks_dir._01_pre_processing_sat_obs._support._n02_funcs import (
    stitch_sentinel_images_tree,
    convert_tree_to_int16,
    clip_raster_tree,
)
```

## Setting directory

```python
from paths.OG_paths import (
    #######################################
    #          GENERAL POLYGONS           #
    #######################################
    # |         Veluwe polygons           |
    # +===================================+
    veluwe_polygon__ruim__UTM32631__gpkg_path,


    #######################################
    #      01_pre_processing_sat_obs      #
    #######################################
    # |     n02_stitch_sat_obs_rstrs      |
    # +===================================+
    # --- sentinel mosaics raw --- #
    s1_mosaics__raw__dir,
    s2_mosaics__raw__dir,

    # --- s2 observations to uint32 --- #
    s2_mosaics__to_unt16__dir,

    # --- sentinel mosaics stiched --- #
    s1_mosaics__stitched__dir,
    s2_mosaics__stitched__dir,

    # --- sentinel mosaics clip --- #
    s1_mosaics__n2000__clip__dir,
    s2_mosaics__n2000__clip__dir,
)
```

## Stitching S1 mosaic data

```python
s1_mosaic_bands = ["VV.tif", "VH.tif"]

stitch_sentinel_images_tree(
    in_root=s1_mosaics__raw__dir, 
    out_root=s1_mosaics__stitched__dir,
    band_filenames=s1_mosaic_bands
)
```

## Clipping S1 mosaic data to N2000

```python
s1_mosaic_clip_var = clip_raster_tree(
    in_root=s1_mosaics__stitched__dir,
    out_root=s1_mosaics__n2000__clip__dir,
    gpkg_path=veluwe_polygon__ruim__UTM32631__gpkg_path,
    pattern="*_stitched.tif",
    nodata=-9999,
    overwrite=True,
    suffix="",
    include_mask_layer=True,
)
print(f"Produced {len(s1_mosaic_clip_var)} clipped rasters")
```

## Stitching S2 mosaic data

```python
s2_mosaic_bands = ["B02.tif", "B03.tif", "B04.tif", "B08.tif", "observations.tif"]

stitch_sentinel_images_tree(
    in_root=s2_mosaics__raw__dir, 
    out_root=s2_mosaics__stitched__dir,
    band_filenames=s2_mosaic_bands
)
```

## Writing S2 data to uint32

```python
written, skipped = convert_tree_to_int16(
    in_root=s2_mosaics__stitched__dir,
    out_root=s2_mosaics__to_unt16__dir,
    pattern="*_stitched.tif",
    overwrite=False,
    nodata=-9999,
    skip_if_int16=False,
)
print("written:", len(written), "skipped:", len(skipped))
```

## Clipping S2 mosaic data to N2000

```python
s2_mosaic_clip_var = clip_raster_tree(
    in_root=s2_mosaics__to_unt16__dir,
    out_root=s2_mosaics__n2000__clip__dir,
    gpkg_path=veluwe_polygon__ruim__UTM32631__gpkg_path,
    pattern="*_stitched.tif",
    nodata=-9999,
    overwrite=True,
    suffix="",
    include_mask_layer=True,
)
print(f"Produced {len(s2_mosaic_clip_var)} clipped rasters")
```
